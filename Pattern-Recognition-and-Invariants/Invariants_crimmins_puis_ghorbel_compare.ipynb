```python
# Définition de la Transformée de Fourier Discrète (DFT)
def dft_from_scratch(x, N=None):
    """
    Calculer la Transformée de Fourier Discrète (DFT) d'un signal x.
    Si N est spécifié, utiliser N coefficients harmoniques.
    """
    if N is None:
        N = len(x)
    n = np.arange(len(x))
    k = np.arange(N).reshape((N, 1))
    # Calcul de la matrice de Fourier avec l'exponentielle complexe
    M = np.exp(-2j * np.pi * k * n / len(x))
    return np.dot(M, x)
```

```python
# Définition de l'inverse de la DFT (iDFT)
def idft_from_scratch(X, N=None):
    """
    Calculer l'inverse de la Transformée de Fourier Discrète (IDFT) d'un signal X.
    Si N est spécifié, utiliser N coefficients harmoniques pour la reconstruction.
    """
    if N is None:
        N = len(X)
    k = np.arange(len(X))
    n = np.arange(N).reshape((N, 1))
    # Calcul de la matrice de Fourier inverse
    M = np.exp(2j * np.pi * k * n / len(X))
    return np.dot(M, X) / len(X)
```

```python
# Invariants de Ghorbel (stables)

def ghorbel_invariants(fourier_coeffs, k0, k1, p=1, q=1):
    """
    Calculate Ghorbel's invariants and modified Fourier coefficients.
    Args:
        fourier_coeffs (ndarray): Fourier coefficients of the curve.
        k0, k1 (int): Indices for calculating invariants.
        p, q (float): Parameters for invariants calculation.
    Returns:
        modified_coeffs (ndarray): Modified Fourier coefficients for reconstruction.
    """
    # Initialize modified coefficients
    N = len(fourier_coeffs)
    modified_coeffs = np.zeros_like(fourier_coeffs, dtype=complex)

    # calcul des invariants
    I_k0 = np.abs(fourier_coeffs[k0])
    I_k1 = np.abs(fourier_coeffs[k1])
    theta0 = np.angle(fourier_coeffs[k0])
    theta1 = np.angle(fourier_coeffs[k1])

    # coefficients invariants modifiés
    modified_coeffs[k0] = (I_k0 ** (1 / (k0 - k1 + p))) * np.exp(1j * 2 * np.pi * theta0)
    modified_coeffs[k1] = (I_k1 ** (1 / (k1 - k0 + q))) * np.exp(1j * 2 * np.pi * theta1)

    # Calculate modified coefficients for k != k0 and k != k1
    for k in range(N):
        if k != k0 and k != k1:
            term1 = I_k0 ** (1 / (k0 - k1))
            term2 = I_k0 ** (-p /  ((k0 - k1)*( k0 - k1 + q)))
            term3 = I_k1 ** (-q / ((k0 - k1) * (k1 - k0 + p)))
            phase_shift = np.exp(((2 * np.pi * 1j / (k0 - k1))) * (k * (theta0 - theta1) - k1*theta0 + k0*theta1))
            modified_coeffs[k] = term1 * term2 * term3 * phase_shift

    return modified_coeffs
```

```python
# Calcul des invariants de Crimmins pour un ensemble donné de coefficients
def crimmins_invariants(C, k0, k1):
    """
    Calculer les invariants de Crimmins I_k pour des indices k0 et k1 donnés.
    """
    invariants = []
    for k in range(len(C)):
        if k == k0:
            I_k = np.abs(C[k0])
        elif k == k1:
            I_k = np.abs(C[k1])
        else:
            I_k = (C[k])**(k0 - k1) * (C[k0])**(k - k1) * (C[k1])**(k - k0)
        invariants.append(I_k)
    return np.array(invariants)
```

```python
from PIL import Image
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
import pandas as pd

# Fonction pour charger une image GIF binaire
def charger_image_gif_binaire(chemin_image):
    """
    Charge une image GIF, convertit en niveaux de gris et applique un seuillage binaire.
    """
    # Charger l'image avec Pillow et convertir en niveaux de gris
    with Image.open(chemin_image) as img:
        img_gray = img.convert("L")  # Convertit en niveaux de gris
    image_np = np.array(img_gray)  # Convertit en array NumPy

    # Appliquer le seuillage binaire
    _, image_binaire = cv2.threshold(image_np, 127, 255, cv2.THRESH_BINARY)
    return image_binaire

# Fonction pour extraire les contours
def extraire_contours(image_binaire):
    """
    Extrait les contours de l'image binaire en utilisant findContours de OpenCV.
    """
    contours, _ = cv2.findContours(image_binaire, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    return contours

# Fonction pour réduire les contours à 60 points
def reduire_contours_a_60_points(contours):
    """
    Réduit les contours extraits à 60 points par sous-échantillonnage.
    """
    # Extraire tous les points des contours
    points_contours = np.vstack([c.reshape(-1, 2) for c in contours])

    # Vérifier qu'il y a suffisamment de points
    if len(points_contours) < 60:
        print("Pas assez de points, réduire le nombre d'objets ou augmenter la résolution.")
        return points_contours

    # Réduire les points à 60 en échantillonnant uniformément
    indices_choisis = np.linspace(0, len(points_contours) - 1, 60, dtype=int)
    points_reduits = points_contours[indices_choisis]
    return points_reduits

# Fonction pour afficher l'image avec les contours trouvés
def afficher_image_avec_contours(image_path, points_reduits):
    """
    Affiche l'image originale avec les contours réduits à 60 points superposés.
    """
    # Charger l'image originale pour l'affichage
    with Image.open(image_path) as img:
        img_rgb = img.convert("RGB")  # Conversion en RGB pour matplotlib
    plt.figure(figsize=(6, 6))
    plt.imshow(img_rgb)
    plt.scatter(points_reduits[:, 0], points_reduits[:, 1], color='red', s=10)
    plt.title("Contours réduits à 60 points")
    plt.gca().invert_yaxis()
    plt.show()

# Fonction pour générer un fichier CSV pour une classe de forme
def generer_csv_points(points_par_image, nom_classe):
    """
    Génère un fichier CSV contenant 60 points (lignes) pour chaque image d'une classe (colonnes).
    """
    for i, points in enumerate(points_par_image):
        df_points = pd.DataFrame(points, columns=['x', 'y'])
        csv_path = f"/content/{nom_classe}_image_{i+1}.csv"
        df_points.to_csv(csv_path, index=False)
        print(f"Fichier {csv_path} créé avec succès.")

# Fonction principale pour traiter les images de toutes les classes et sauvegarder les résultats
def traiter_images_par_classe(classe_paths, nom_classe):
    """
    Traite les images d'une classe spécifique, extrait et affiche les contours,
    puis sauvegarde les points dans des fichiers CSV.
    """
    points_par_image = []

    # Parcourir chaque image dans le dossier de la classe
    for image_name in os.listdir(classe_paths):
        image_path = os.path.join(classe_paths, image_name)
        image_binaire = charger_image_gif_binaire(image_path)
        contours = extraire_contours(image_binaire)
        points_reduits = reduire_contours_a_60_points(contours)

        # Stocker les points pour le fichier CSV
        points_par_image.append(points_reduits)

        # Afficher les contours sur l'image
        afficher_image_avec_contours(image_path, points_reduits)

    # Générer un fichier CSV pour chaque image de la classe
    generer_csv_points(points_par_image, nom_classe)

# Dictionnaire des chemins d'accès pour chaque classe
chemins_classes = {
    "C1_Coeur": "/content/drive/MyDrive/Workshop_RF_Data/C1_Coeur",
    "C2_Arbre": "/content/drive/MyDrive/Workshop_RF_Data/C2_Arbre",
    "C3_Ours": "/content/drive/MyDrive/Workshop_RF_Data/C3_Ours",
    "C4_Voiture": "/content/drive/MyDrive/Workshop_RF_Data/C4_Voiture",
    "C5_Marteau": "/content/drive/MyDrive/Workshop_RF_Data/C5_Marteau"
}

# Traiter chaque classe et sauvegarder les points extraits
for nom_classe, chemin_classe in chemins_classes.items():
    print(f"Traitement de la classe : {nom_classe}")
    traiter_images_par_classe(chemin_classe, nom_classe)
```

```python
from PIL import Image
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
import pandas as pd

# Fonction pour charger une image GIF binaire
def charger_image_gif_binaire(chemin_image, est_marteau=False):
    """
    Charge une image GIF, convertit en niveaux de gris et applique un seuillage binaire.
    Si 'est_marteau' est True, un seuil plus sensible est appliqué pour le marteau.
    """
    with Image.open(chemin_image) as img:
        img_gray = img.convert("L")
    image_np = np.array(img_gray)

    # Appliquer un seuil plus sensible pour le marteau
    seuil = 100 if est_marteau else 127
    _, image_binaire = cv2.threshold(image_np, seuil, 255, cv2.THRESH_BINARY)
    return image_binaire

# Fonction pour extraire les contours
def extraire_contours(image_binaire):
    contours, _ = cv2.findContours(image_binaire, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    return contours

# Fonction pour réduire les contours à 60 points
def reduire_contours_a_60_points(contours):
    points_contours = np.vstack([c.reshape(-1, 2) for c in contours])
    if len(points_contours) < 60:
        print("Pas assez de points pour échantillonner à 60.")
        return points_contours
    indices_choisis = np.linspace(0, len(points_contours) - 1, 60, dtype=int)
    points_reduits = points_contours[indices_choisis]
    return points_reduits

# Fonction pour générer un fichier CSV unique pour chaque classe
def generer_csv_par_classe(points_par_image, noms_images, nom_classe):
    """
    Génère un fichier CSV unique pour la classe spécifiée, avec des colonnes pour chaque image.
    """
    # Créer un DataFrame pour stocker les données des points
    df_points = pd.DataFrame()

    # Ajouter chaque paire de colonnes pour chaque image
    for i, points in enumerate(points_par_image):
        # Nommer les colonnes avec le nom de l'image
        nom_col_x = f"{noms_images[i]}_x"
        nom_col_y = f"{noms_images[i]}_y"
        df_points[nom_col_x] = points[:, 0]
        df_points[nom_col_y] = points[:, 1]

    # Enregistrer le DataFrame dans un fichier CSV
    csv_path = f"/content/{nom_classe}_contours.csv"
    df_points.to_csv(csv_path, index=False)
    print(f"Fichier {csv_path} créé avec succès.")

# Fonction principale pour traiter les images d'une classe
def traiter_images_par_classe(classe_paths, nom_classe, est_marteau=False):
    """
    Traite les images d'une classe spécifique, extrait les contours,
    puis sauvegarde tous les points dans un seul fichier CSV pour la classe.
    """
    points_par_image = []
    noms_images = []

    for image_name in os.listdir(classe_paths):
        image_path = os.path.join(classe_paths, image_name)
        image_binaire = charger_image_gif_binaire(image_path, est_marteau)
        contours = extraire_contours(image_binaire)
        points_reduits = reduire_contours_a_60_points(contours)

        # Ajouter les points réduits et le nom de l'image pour le fichier CSV
        points_par_image.append(points_reduits)
        noms_images.append(os.path.splitext(image_name)[0])  # Nom sans extension

    # Générer un fichier CSV avec les points pour la classe entière
    generer_csv_par_classe(points_par_image, noms_images, nom_classe)

# Dictionnaire des chemins d'accès pour chaque classe
chemins_classes = {
    "C1_Coeur": "/content/drive/MyDrive/Workshop_RF_Data/C1_Coeur",
    "C2_Arbre": "/content/drive/MyDrive/Workshop_RF_Data/C2_Arbre",
    "C3_Ours": "/content/drive/MyDrive/Workshop_RF_Data/C3_Ours",
    "C4_Voiture": "/content/drive/MyDrive/Workshop_RF_Data/C4_Voiture",
    "C5_Marteau": "/content/drive/MyDrive/Workshop_RF_Data/C5_Marteau"
}

# Traiter chaque classe, avec une sensibilité ajustée pour "C5_Marteau"
for nom_classe, chemin_classe in chemins_classes.items():
    est_marteau = (nom_classe == "C5_Marteau")
    print(f"Traitement de la classe : {nom_classe}")
    traiter_images_par_classe(chemin_classe, nom_classe, est_marteau)
```
